# Choose a model

```python
MODEL_ID = "gemini-3-flash-preview"
```

# Send text prompt

interactions.create -> method to generate responses from text-only input

```python
interaction = client.interactions.create(
    model=MODEL_ID,
    input="Explain how AI works in a few words",
)

# The response is an Interaction object containing a list of steps.
# Each step represents a part of the model's response.
# Let's look at what the steps contain:
print(f"Number of steps: {len(interaction.steps)}")
for j, step in enumerate(interaction.steps):
    print(f"  Step {j}: type={step.type}")
```

```python
interaction.output_text
```

## Some useful parameters in create method

**system_instruction** – System instruction for the interaction to steer the behavior of the model. Note that system instructions will be set for the whole chain of interaction and not per turn.*kursywa*

```python
system_instruction = "You are Ayn Rand and are explaining things to 40 years old woman."

interaction_1 = client.interactions.create(
    model=MODEL_ID,
    input="Explain how AI works",
    system_instruction=system_instruction,
)

Markdown(interaction_1.output_text)
```

**previous_interaction_id** - Gemini API enables multi-turn conversations using previous_interaction_id to chain interactions. The model remembers the full conversation context — you just pass the ID from the previous turn and all the history of the previous turns will be kept.

```python
interaction_2 = client.interactions.create(
    model=MODEL_ID,
    input="Interesting! Is AI a new engine that needs to be protected from looters?",
    previous_interaction_id=interaction_1.id,
)

Markdown(interaction_2.output_text)
```

generation_config - bag of parameters with we can use to modify out response from model

*   **`temperature`**: Controls the randomness of the output. A higher temperature (e.g., 1.0) results in more creative and diverse responses, while a lower temperature (e.g., 0.2) makes the output more predictable and focused. The valid range is (0.0, 2.0].
*   **`top_p`**: The maximum cumulative probability of tokens to consider. The model samples from the most likely tokens whose cumulative probability exceeds `top_p`. For example, a `top_p` of 0.9 means the model considers the smallest set of most likely tokens whose cumulative probability sums to 0.9 or greater. This helps to avoid generating low-probability tokens.
*   **`top_k`**: The maximum number of tokens to consider when sampling. The model samples from the `top_k` most likely tokens. This also helps to control the diversity of the output.
*   **`max_output_tokens`**: The maximum number of tokens to generate in the response. This is useful for controlling the length of the generated output.

```python
interaction = client.interactions.create(
    model=MODEL_ID,
    input="Tell me how the internet works, but pretend I'm a kid who understands only dog-related analogies.",
    generation_config={
    "temperature": 0.7,
    "top_p": 0.9,
    "top_k": 40,
    "max_output_tokens": 5000,
    },
)

Markdown(interaction.output_text)
```
